## 1. Dos corridas

**a)** Si la pérdida de entrenamiento disminuye mientras la de validación aumenta, hay sobreajuste. Si ambas permanecen altas o casi constantes, el modelo no está aprendiendo adecuadamente.

**b)** Para sobreajuste, aplicar regularización o *early stopping*; funcionaría si la pérdida de validación deja de aumentar. Si no aprende, revisar la tasa de aprendizaje y la inicialización; funcionaría si la pérdida disminuye de forma sostenida.

## 2. Un cálculo de varianza

**a)** El factor por capa es:

\[
\frac{n_{in}\sigma^2}{2}=\frac{400(0.05)^2}{2}=0.5
\]

**b)** Después de 10 capas:

\[
\frac{\operatorname{std}(a^{10})}{\operatorname{std}(a^0)}
=\sqrt{0.5^{10}}=0.03125
\]

**c)** Para conservar la varianza:

\[
\sigma=\sqrt{\frac{2}{400}}\approx0.0707
\]

Se denomina inicialización He o Kaiming.

## 3. Tres inicializaciones

**a)** II corresponde a Xavier por la rápida atenuación de las activaciones; III presenta una caída progresiva de la desviación estándar; I mantiene activaciones cercanas a 1, pero presenta normas de gradiente elevadas. La asignación exacta entre PyTorch por defecto y He no puede establecerse con certeza solo con las tendencias descritas.

**b)** Elegiría He para una red con ReLU, porque busca conservar la escala de las activaciones y facilitar la propagación de gradientes.

## 4. La pérdida inicial

**a) 1.95:** valor normal, pues \(\ln(7)\approx1.946\), la entropía cruzada esperada para predicciones uniformes en siete clases.

**b) 38.2:** valor excesivamente alto; sospecharía predicciones muy seguras y equivocadas, logits extremos o un problema numérico/de implementación.

**c) 0.03:** valor inusualmente bajo antes de entrenar; revisaría fuga de información, evaluación incorrecta o si el modelo ya fue entrenado.

## 5. Tamaño del lote

**a)** Q usa lotes de 16: la curva es más ruidosa y tarda 0.82 s por época. P usa lotes de 1024: la curva es más suave y tarda 0.14 s por época.

**b)** La regla de escalado lineal daría \(\eta_{nuevo}=0.05(1024/16)=3.2\), aunque debe validarse experimentalmente porque puede ser inestable. El *warmup* incrementa gradualmente la tasa de aprendizaje para evitar actualizaciones iniciales demasiado grandes.

## 6. Profundidad y weight decay

**a)** No hay evidencia clara de sobreajuste en la red de 40 capas: las pérdidas de entrenamiento y validación terminan aproximadamente en 0.53 y 0.57. Probaría entrenar más épocas y revisar la tasa de aprendizaje antes de añadir Dropout.

**b)** Adam tradicional incorpora la penalización en el gradiente antes de la adaptación de Adam. AdamW desacopla el *weight decay* de la actualización adaptativa, lo que permite una regularización más consistente.
